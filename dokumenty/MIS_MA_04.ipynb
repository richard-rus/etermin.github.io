Text

Podľa štúdie publikovanej v odbornom časopise Journal of Medical Internet Research (Impact of Online Appointment Scheduling, https://www.jmir.org/2017/4/e134/) dochádza po zavedení online objednávacieho systému k výraznému zníženiu miery vymeškania termínov.

Pri tradičnom telefonickom objednávaní bola miera neúčasti 17,5 %, zatiaľ čo pri online systéme len 10,1 %. To predstavuje pokles o viac ako 42 %. Pre ambulanciu to znamená lepšie využitie kapacít, menej prázdnych termínov a vyššie príjmy. Rovnako významná je úspora času. Telefonická rezervácia trvá v priemere približne 480 sekúnd, zatiaľ čo online rezervácia len 105 sekúnd. Ide o takmer 78 % úsporu času pre pacienta a zároveň výrazné zníženie administratívnej záťaže personálu. Online systém je navyše dostupný nepretržite – 24 hodín denne, 7 dní v týždni. Telefónne objednávanie je obmedzené na pracovné hodiny ambulancie približne 8 až 9 hodín denne. To znamená približne trojnásobnú dostupnosť služby. Dôležitým faktorom je aj skrátenie predstihu objednania. Pri manuálnom systéme bol priemerný predstih 14,2 dňa, pri online systéme 8,4 dňa. Kratší čas medzi objednaním a termínom zvyšuje flexibilitu a znižuje riziko, že pacient na termín zabudne.

Pre náš projekt eTermin tieto výsledky predstavujú silný argument, že investícia do digitálnej aplikácie na objednávanie sa k lekárovi je nielen technologickým krokom vpred, ale aj ekonomicky a prevádzkovo výhodným riešením pre zdravotnícke zariadenia.

Prezentácia

Prečo sa digitálne objednávanie oplatí?
- Preukázateľné zlepšenie efektivity ambulancií
- Menej vymeškaných termínov
- Úspora času pre pacientov aj personál
- Vyššia dostupnosť služby
- Lepšie využitie kapacít lekára

- Telefón: 17,5 %
- Online: 10,1 %
- Pokles o 42 %

Čas rezervácie
- Telefón: ~480 s
- Online: ~105 s

- Telefón: 8–9 h denne

- 300 % vyššia dostupnosť

Predstih objednania
- Telefón: 14,2 dňa
- Online: 8,4 dňa
- Skrátenie o 59 %

```python
from matplotlib import pyplot as plt
import pandas as pd

data = {
    "Metrika": [
        "Miera vymeškania termínov",
        "Čas na rezerváciu",
        "Dostupnosť (v hodinách)",
        "Predstih objednania"
    ],
    "Manuálny systém (Telefón)": ["17,5 %", "~480 sekúnd", "8 – 9 hod / denne", "14,2 dňa"],
    "Online aplikácia": ["10,1 %", "~105 sekúnd", "24 / 7", "8,4 dňa"],
    "Zmena (Efektivita)": [
        "-42,3 %",
        "-78 %",
        "+300 %",
        "-59,2 %"
    ]
}

df = pd.DataFrame(data)

plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Helvetica Neue', 'Helvetica', 'Arial', 'Segoe UI', 'sans-serif']

fig, ax = plt.subplots(figsize=(14, 4))
fig.patch.set_alpha(0)
ax.axis('tight')
ax.axis('off')

table = ax.table(
    cellText=df.values,
    colLabels=df.columns,
    cellLoc='center',
    loc='center'
)

table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.2, 3)

for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_text_props(weight='bold')
        cell.set_facecolor('#dce6f2')

plt.savefig('MIS_TABLE_COMPARISON', bbox_inches='tight', dpi=300, transparent=True)
plt.show()
```

```python
import numpy as np

numeric_data = [
    {"title": "Miera vymeškania termínov", "manual": 17.5, "online": 10.1, "unit": "%",      "filename": "01_miera_vymeskania.png"},
    {"title": "Čas na rezerváciu",          "manual": 480,  "online": 105,  "unit": "sekúnd", "filename": "02_cas_na_rezervaciu.png"},
    {"title": "Dostupnosť (v hodinách)",     "manual": 8.5,  "online": 24,   "unit": "hodín",  "filename": "03_dostupnost.png"},
    {"title": "Predstih objednania",                    "manual": 14.2, "online": 8.4,  "unit": "dní",    "filename": "04_predstih.png"},
]

plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Helvetica Neue', 'Helvetica', 'Arial', 'Segoe UI', 'sans-serif']

for m in numeric_data:
    fig, ax = plt.subplots(figsize=(7, 4.5))
    fig.patch.set_alpha(0)
    ax.set_facecolor('none')

    categories = ['Manuálny systém (Telefón)', 'Online aplikácia']
    values = [m['manual'], m['online']]
    x = np.arange(len(categories))

    ax.bar(x, values, width=0.45, color=['#4caf82', '#1f77b4'])

    ax.set_title(m['title'], weight='bold')

    ax.set_xticks(x)
    ax.set_xticklabels(categories)

    ax.set_ylabel(m['unit'])

    plt.savefig(m['filename'], bbox_inches='tight', dpi=300, transparent=True)
    plt.show()
```
